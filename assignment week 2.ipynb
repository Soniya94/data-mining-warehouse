```python
import numpy as np
import pandas as pd

np.random.seed(42)

# Dataset size
n = 500

# Dimensions
products = ['TV', 'Laptop', 'Phone']
cities = ['London', 'Manchester', 'Birmingham', 'Leeds']
countries = ['UK']

# Generate data
data = {
    "Product": np.random.choice(products, n),
    "City": np.random.choice(cities, n),
    "Country": np.random.choice(countries, n),
    "Season": np.random.choice(['Q1', 'Q2', 'Q3', 'Q4'], n),
    "Date": pd.date_range(start='2023-01-01', periods=n, freq='D'),
    "Quantity": np.random.randint(1, 20, n),
    "Price": np.random.randint(200, 2000, n)
}

df = pd.DataFrame(data)

# Save files
df.to_csv("electronics_sales.csv", index=False)
df.to_json("electronics_sales.json", orient="records")

df.head(10)
```
